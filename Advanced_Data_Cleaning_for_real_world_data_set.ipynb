```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
DATA = pd.read_csv(r"https://raw.githubusercontent.com/KeithGalli/Olympics-Dataset/refs/heads/master/athletes/bios.csv")
```

# What to clean from this data set by observing the data
- [x] Get ride of bullet in used names
- [x] Get ride of columns not usefull
- [x] split 'height'/'wight' and convert them to integers
- [x] parse out date from 'Born' and 'Died' column
- [x] parse out city, country and region from 'Born'
- [x] calculate age at death

```python
df = DATA.copy()
```

```python
df.head(1)
```

```python
df.info()
```

```python

# does all the used name contains that bullet point in between
len(df)
```

```python
# so all the names in the df contains "•"
len(df[df['Used name'].str.contains("•")])
```

```python
# remove that character from the used name
df['name'] = df['Used name'].str.replace("•"," ")
```

```python
len(df[~df.Measurements.isna()])
```

```python
# make sure all the measurment columns contain "/" unless we have to use another apporach to split measurements wight and height
df[~df["Measurements"].str.contains("/",na=False) & df["Measurements"].notna()]
```

```python
df.columns = df.columns.str.lower().str.strip().str.replace(" ",'_')
df.head(5)
```

```python
# we will creat a toy data frame to check our codes on small scale and then to ther real project
toy = pd.DataFrame(['100 cm','125 cm/45 kg', '35 cm/25 kg','100 kg'], columns = ['measurements'])
toy
```

```python
toy['height'] = toy['measurements'].str.extract(r"(\d+\s*cm)")
toy['weight'] = toy['measurements'].str.extract(r"(\d+\s*kg)")
toy
```

```python
# we applied the code what we have tested on the small scale data
df['height_cm'] = df.measurements.str.extract(r"(\d+\s*cm)")
df['weight_kg'] = df.measurements.str.extract(r"(\d+\s*kg)")
df['height_cm'] = df['height_cm'].str.replace("cm","").astype('Int64')
df['weight_kg'] = df['weight_kg'].str.replace("kg","").astype('Int64')
df.head(2)
```

```python
# check again for changes that is made and spliting the measurements found to be good with the extract method
df[~df["measurements"].str.contains("/",na=False) & df["measurements"].notna()].head()
```

```python
# who is the tallest athlet
df.sort_values('height_cm',ascending=False).head(1)
```

```python
# extract the date formate from 'born' column
pattern = r'(\d+ \w+ \d{4})'
pattern2 = r'(\d{1,2} \w+ \d{4}|\d{4})'
df['born_date'] = df['born'].str.extract(r'(\d{1,2} \w+ \d{4}|\d{4})')
df['date_died'] = df['died'].str.extract(r'(\d{1,2} \w+ \d{4}|\d{4})')

```

```python
# change both the new columns born and died date to datetime data type
df['born_date'] = pd.to_datetime(df['born_date'],format="mixed")
df['date_died'] = pd.to_datetime(df['date_died'],format="mixed")
```

```python
# here we realize that the born column is not uniform as we expected and the above method will not be usefull
df[~df['born'].str.match(pattern,na=False) & df['born'].notna()]
```

```python
# check for the above code if it works correctly by filttering the data
df[~df['born'].str.match(pattern,na=False) & df['born'].notna() & ~df['born'].str.match(pattern2,na=False)]
```

```python
df.info()
```

```python
#df['age_at_death1'] = (df['date_died'] - df['born_date'])/np.timedelta64(1,'Y')
df['age_at_death'] = ((df['date_died'] - df['born_date']).dt.days/360).round().astype('Int64')
```

```python
#df[['city','region','country',]] = df['born'].extract(r" in (\w+)")
#df[['city','region']] = df['born'].str.extract(r"in (\w+)\,+(\w+)")
df[['city', 'region', 'country']] = df['born'].str.extract(r'in\s+([^,]+),\s*([^(]+)\s*\(([^)]+)\)')

df.columns
```

```python
data_cleaned = df.drop(columns = ['used_name', 'born','noc','athlete_id', 'measurements', 'affiliations', 'nick/petnames','title(s)', 'other_names', 'nationality', 'original_name', 'name_order','name'])
data_cleaned.head()
```
