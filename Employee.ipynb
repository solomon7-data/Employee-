```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#. perform EDA
df = pd.read_excel(r"C:\Users\solomon\Desktop\employee.xlsx")
```

```python
df.head()
```

```python
df.info()
```

```python
# calculate the percent of missing value for each columns
df.isna().sum()
```

```python
df['Age'] = pd.to_numeric(df['Age'],errors = 'coerce')
df['Experience'] = pd.to_numeric(df['Experience'],errors = 'coerce')
```

```python
df.info()
```

```python
#3  Fill missing ages with the median age. 
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
#4  Fill missing salaries with the average salary. 
df['Salary'] = df['Salary'].fillna(df['Salary'].mean()).round(1)
```

```python
#5  Fill missing cities with "Unknown".
df['City'] = df['City'].fillna('Unknown')
```

```python
df['Join_Date'] = pd.to_datetime(df['Join_Date'],errors = 'coerce')
```

```python
#6  Convert Join_Date to datetime.
df['Join_Date'].dtype
```

```python
# Check whether there are any duplicates and remove them
df.loc[df.duplicated()]
```

```python
df = df.drop_duplicates()
```

```python
# Check whether there are any duplicates.
df.loc[df.duplicated()]
```

```python
#Count employees in each city.
df.City.value_counts()
```

```python
#Find the average experience by department.
df.groupby('Department').Experience.mean()
```

```python
# Create a column called Bonus equal to 10% of salary.
df['bonus'] = (df['Salary']*0.1).round(1)
```

```python
df.head()
```

```python
# Create a column senior with  a label yes and no
df['Senior'] = np.where(df['Experience'] >=5,'Yes','No')
```

```python
df.head()
```

```python
df['Year'] = df.Join_Date.dt.year
```

```python
df.head()
```

```python
# Find employees who joined after 2020.
df[df.Year > 2020]
```

```python
#Find the top 5 highest-paid employees.
df.sort_values(by = 'Salary',ascending=False).head(5)
```

```python
#Find employees whose names start with "A"
df[df.Name.str.startswith('A',na=False)]
```

```python
# Find employees whose names contain the letter "a"
df[df.Name.str.contains('a')]
```
